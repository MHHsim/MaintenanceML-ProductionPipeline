# Fase 5: Resultados

## 5.1 Resumen del proceso

- **Datos:** el dataset original tenía 10.000 registros y 14 columnas, sin valores nulos ni filas duplicadas.
- **Limpieza:** se eliminaron los identificadores `UDI` y `Product ID`, las 19 filas con `RNF = 1` y las columnas de tipo de fallo (`TWF`, `HDF`, `PWF`, `OSF` y `RNF`), porque revelaban casi por completo el valor de `Machine failure` (coincidencia en el 99,7 % de las filas) y solo se conocen cuando el fallo ya ha ocurrido. El dataset limpio tiene 9.981 filas y 7 columnas.

## 5.2 Principales hallazgos

- **Desbalanceo:** hay 9.643 registros sin fallo (96,61 %) frente a 338 con fallo (3,39 %).
- **Distribuciones:** solo `Rotational speed [rpm]` es claramente asimétrica (sesgo de 1,993). Ella y `Torque [Nm]` son las únicas con valores atípicos según el criterio IQR.
- **`Type`:** la tasa de fallo es más alta en la calidad baja (L, 3,91 %) que en la media (M, 2,77 %) y la alta (H, 2,10 %). La asociación es significativa (chi-cuadrado, p = 0,0012), aunque la diferencia es moderada.
- **Variables numéricas:** las cinco difieren de forma significativa entre los registros con fallo y sin fallo (Mann-Whitney, p < 0,05). Las que más separan los grupos son `Torque [Nm]`, con valores más altos en los fallos, y `Rotational speed [rpm]`, con valores más bajos. Les siguen `Tool wear [min]` y `Air temperature [K]`, y la que menos, `Process temperature [K]`.

## 5.3 Relación con el objetivo de negocio

El análisis identifica qué variables se asocian con los fallos, que era el objetivo analítico planteado: torque, velocidad de rotación y desgaste de la herramienta son las más relacionadas. Se trata de asociaciones observadas, no de relaciones causales.

## 5.4 Limitaciones

- El dataset es sintético, por lo que puede no reflejar del todo un entorno real.
- Hay pocos fallos (338), por lo que las conclusiones sobre ese grupo son menos estables.
- Con casi 10.000 registros, los p-valores son muy pequeños incluso con diferencias modestas, así que la significación estadística no implica por sí sola una relación fuerte.
