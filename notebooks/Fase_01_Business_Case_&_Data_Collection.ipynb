# Fase 1: Business Case & Data Collection

## 1.1 Business Case

### Contexto
En un proceso de producción industrial, las máquinas de fresado están expuestas a desgaste, sobrecalentamiento y sobreesfuerzo. Cuando una máquina falla de forma inesperada, se detiene la producción.

### Problema
Los fallos inesperados generan paradas no planificadas, costes de reparación y retrasos en la producción. El mantenimiento reactivo (arreglar cuando ya ha fallado) es más caro que el preventivo.

### Objetivo de negocio
Anticipar los fallos de las máquinas para poder programar el mantenimiento antes de que ocurran, reduciendo paradas y costes.

### Objetivo analítico
Analizar los datos de sensores del proceso para entender qué condiciones (temperatura, velocidad, torque, desgaste) se asocian con los fallos, y preparar los datos para un futuro modelo que prediga `Machine failure` (0 = no falla, 1 = falla).

### Criterios de éxito
- Identificar qué variables influyen más en los fallos.
- Dejar un dataset limpio y sin fuga de datos, listo para modelar.

### Limitaciones
- El dataset es sintético, por lo que puede no reflejar del todo un entorno real.
- Es probable que los fallos sean poco frecuentes frente al funcionamiento normal, lo que se comprobará en el análisis.
